# 04: Multi-Table Data Integration & Staging Resolution
## Recruitflow Entity Relationship Reconstruction

### Objectives
Reconstruct candidate journeys by integrating four core tables: `candidates`, `jobs`, `applications`, and `stage_events`.


```python
import pandas as pd
import plotly.express as px

df_cand = pd.DataFrame([{'cand_id': 'C-1', 'name': 'Samir Patel'}, {'cand_id': 'C-2', 'name': 'Sarah Jenkins'}])
df_apps = pd.DataFrame([{'app_id': 'A-101', 'cand_id': 'C-1', 'dept': 'Engineering'}, {'app_id': 'A-102', 'cand_id': 'C-2', 'dept': 'Design'}])
df_events = pd.DataFrame([
    {'app_id': 'A-101', 'stage': 'Screening', 'outcome': 'passed'},
    {'app_id': 'A-101', 'stage': 'Interview', 'outcome': 'dropped'},
    {'app_id': 'A-102', 'stage': 'Screening', 'outcome': 'passed'},
    {'app_id': 'A-102', 'stage': 'Joined', 'outcome': 'passed'}
])

journey_joined = df_events.merge(df_apps, on='app_id').merge(df_cand, on='cand_id')
display(journey_joined)
```
