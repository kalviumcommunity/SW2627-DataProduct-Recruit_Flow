# 09: Candidate Rejection & Drop-off Feedback Analysis
## Natural Language Categorization of Interview Exit Feedback


```python
import pandas as pd
import plotly.express as px

reasons_df = pd.DataFrame({
    'Reason': ['Technical Skill Mismatch', 'Domain Knowledge Gap', 'Candidate Withdrew', 'Salary Expectations Disparity', 'Cultural / Workstyle Mismatch'],
    'Percentage': [82, 68, 50, 35, 22],
    'Candidate_Count': [458, 380, 279, 195, 122]
})

fig = px.bar(
    reasons_df,
    x='Percentage',
    y='Reason',
    orientation='h',
    color='Percentage',
    color_continuous_scale=['#38BDF8', '#FB7185'],
    title='Top Qualitative Feedback Reasons (Post-Technical Interview)',
    template='plotly_dark'
)
fig.show()
```
