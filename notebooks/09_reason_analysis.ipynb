# Step 9: Drop-off Reason Analysis

This notebook performs drop-off reason analytics to:
1. Aggregate and rank candidate rejection/drop-off reasons.
2. Cross-tabulate drop-off reasons by recruitment stage.
3. Cross-tabulate drop-off reasons by department.

```python
import os
import pandas as pd
import numpy as np

# Path setup
FEATURES_PATH = os.path.join("..", "data", "processed", "candidate_features.csv")
print(f"Checking features file at: {FEATURES_PATH}")
```

## Load Data

We load the engineered candidate features dataset.

```python
df = pd.read_csv(FEATURES_PATH)
df.head()
```

## Part 1: Aggregated Drop-off Reasons

We aggregate and calculate candidate counts and percentages for each rejection/drop-off reason.

```python
# Filter for candidates who dropped
df_dropped = df[df['dropped'] == 1].copy()
df_dropped['rejection_reason'] = df_dropped['rejection_reason'].fillna('Unspecified Drop-off')

total_dropped = len(df_dropped)
print(f"Total Dropped Candidates: {total_dropped}\n")

# Aggregate counts
reasons_df = df_dropped['rejection_reason'].value_counts().reset_index(name='count')
reasons_df.columns = ['reason', 'count']
reasons_df['value'] = reasons_df['count']
reasons_df['percentage'] = round((reasons_df['count'] / total_dropped) * 100.0, 2)

reasons_df
```

## Part 2: Cross-tabulations

We cross-tabulate drop-off reasons across different recruitment stages and departments to isolate where specific friction points occur.

```python
# Rename column for clarity
df_dropped_renamed = df_dropped.rename(columns={'furthest_stage_reached': 'stage'})

print("--- Rejection Reasons by Department ---")
dept_ct = pd.crosstab(df_dropped_renamed['department'], df_dropped_renamed['rejection_reason'])
display(dept_ct)

print("\n--- Rejection Reasons by Recruitment Stage ---")
stage_ct = pd.crosstab(df_dropped_renamed['stage'], df_dropped_renamed['rejection_reason'])
display(stage_ct)
```
