# 03: Data Validation, Schema Enforcement & Quarantine Pipeline
## Recruitflow Automated Data Quality Assurance

### Objectives
1. Implement rigorous validation assertions (unique candidate keys, valid emails, chronologically monotonic stage events).
2. Divert malformed or corrupted records into the **Core Quarantine Error Table** without aborting batch ingestion.


```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

test_batch = pd.DataFrame([
    {'id': 'APP-01', 'email': 'valid.user@acme.com', 'stage_order': 1, 'entered_at': '2026-01-01', 'status': 'valid'},
    {'id': 'APP-02', 'email': 'missing_at_symbol', 'stage_order': 2, 'entered_at': '2026-01-05', 'status': 'quarantine'},
    {'id': 'APP-03', 'email': 'candidate3@test.io', 'stage_order': -1, 'entered_at': '2026-01-10', 'status': 'quarantine'},
    {'id': 'APP-04', 'email': 'candidate4@test.io', 'stage_order': 3, 'entered_at': '2026-01-12', 'status': 'valid'}
])

validation_counts = test_batch['status'].value_counts()

fig = go.Figure(data=[go.Pie(
    labels=validation_counts.index, 
    values=validation_counts.values, 
    hole=.5,
    marker_colors=['#22C55E', '#EF4444']
)])
fig.update_layout(title='Batch Ingestion Validation Yield (%)', template='plotly_dark')
fig.show()
```
