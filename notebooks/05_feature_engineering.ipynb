## Derived Variables, Latency Ratios & Drop-off Flags

### Objectives
Compute core analytical features:
- `duration_days`: Time elapsed in each stage.
- `dropoff_flag`: Binary indicator (True if dropped at stage).
- `velocity_ratio`: Ratio of candidate stage time relative to department median.


```python
import pandas as pd
import numpy as np
import plotly.express as px

np.random.seed(42)
n = 1000
df_feat = pd.DataFrame({
    'application_id': [f'APP-{i}' for i in range(n)],
    'total_stages_entered': np.random.randint(1, 6, n),
    'duration_screening_days': np.random.gamma(2, 2, n).round(1),
    'duration_interview_days': np.random.gamma(3, 3, n).round(1),
    'has_dropoff': np.random.choice([True, False], n, p=[0.78, 0.22])
})
df_feat['total_pipeline_days'] = df_feat['duration_screening_days'] + df_feat['duration_interview_days']
df_feat.head()
```
