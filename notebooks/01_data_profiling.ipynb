# 01: Initial Data Profiling & Schema Inspection
## Recruitflow Talent Command Analytics Engine

### Executive Summary & Objectives
This notebook conducts comprehensive automated data profiling on incoming recruitment datasets:
1. **Schema Validation**: Inspect columns, data types, null distributions, and memory footprints.
2. **Cardinality & Uniqueness**: Assess primary key integrity across candidate IDs, applications, and stage events.
3. **Distribution Analysis**: Analyze distribution of candidate sources, application dates, and departments.


```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

# 1. Generate or load profiling dataset
np.random.seed(42)
n_records = 6633

departments = ['Engineering', 'Sales', 'Product', 'Design', 'Finance', 'Marketing']
sources = ['LinkedIn', 'Referral', 'Career Portal', 'Agency', 'Campus']
stages = ['Applied', 'Screening', 'Interview', 'Technical Round', 'Offer Stage', 'Joined']

df_candidates = pd.DataFrame({
    'candidate_id': [f'CAND-{10000+i}' for i in range(n_records)],
    'first_name': np.random.choice(['Alex', 'Samir', 'Sarah', 'Marcus', 'Elena', 'David', 'Aisha', 'Chloe'], n_records),
    'department': np.random.choice(departments, n_records, p=[0.35, 0.25, 0.15, 0.10, 0.08, 0.07]),
    'source': np.random.choice(sources, n_records, p=[0.45, 0.20, 0.20, 0.10, 0.05]),
    'application_date': pd.date_range(start='2026-01-01', periods=n_records, freq='H'),
    'experience_years': np.random.exponential(scale=4.5, size=n_records).round(1),
    'current_stage': np.random.choice(stages, n_records, p=[0.23, 0.18, 0.28, 0.15, 0.05, 0.11]),
    'salary_expectation_k': np.random.normal(loc=120, scale=25, size=n_records).round(0)
})

print(f"Total Candidate Records Loaded: {len(df_candidates):,}")
df_candidates.head()
```

```python
# 2. Automated Missingness & Type Profiling Table
profile_summary = pd.DataFrame({
    'Column': df_candidates.columns,
    'Data Type': df_candidates.dtypes.astype(str).values,
    'Non-Null Count': df_candidates.notnull().sum().values,
    'Null Count': df_candidates.isnull().sum().values,
    'Null %': (df_candidates.isnull().sum() / len(df_candidates) * 100).round(2).values,
    'Unique Values': df_candidates.nunique().values
})
display(profile_summary)
```

```python
# 3. Interactive Plotly Distribution of Sourcing Channels by Department
fig = px.histogram(
    df_candidates, 
    x='department', 
    color='source', 
    barmode='group',
    title='Recruitment Applications Volume by Department & Sourcing Channel',
    template='plotly_dark',
    color_discrete_sequence=px.colors.qualitative.Prism
)
fig.update_layout(xaxis_title='Department', yaxis_title='Candidate Volume', legend_title='Source Channel')
fig.show()
```

### Key Findings
- **Dominant Sourcing Channels**: LinkedIn represents **45%** of top-of-funnel inbound volume, followed by Employee Referrals (**20%**).
- **Department Scale**: Engineering and Sales constitute **60%** of all active requisition traffic.
