# 07: Stage-Wise Drop-off Root Cause & Bottleneck Exploration
## Detailed Analysis of the Interview Stage Leak


```python
import pandas as pd
import plotly.express as px

dropoff_stages = pd.DataFrame({
    'Stage': ['Applications → Screening', 'Screening → Interview', 'Interview → Technical (Leak)', 'Technical → Offer', 'Offer → Joined (Invisible Loss)'],
    'Lost_Candidates': [1528, 919, 1185, 1310, 221],
    'Severity': ['Medium', 'Low', 'Critical Leak', 'Medium', 'Invisible Loss']
})

fig = px.bar(
    dropoff_stages,
    x='Stage',
    y='Lost_Candidates',
    color='Severity',
    color_discrete_map={'Critical Leak': '#EF4444', 'Medium': '#F59E0B', 'Low': '#38BDF8', 'Invisible Loss': '#F43F5E'},
    title='Lost Candidate Volume by Pipeline Transition Stage',
    template='plotly_dark'
)
fig.show()
```
