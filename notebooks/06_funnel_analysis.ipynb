# 06: End-to-End Recruitment Funnel Conversion Analytics
## Recruitflow Talent Command Pipeline Analysis

### Funnel Metrics Overview
- **Total Applications**: 6,633
- **Screening**: 5,105
- **Interview (Biggest Leak)**: 4,186 (28.9% Drop-off)
- **Technical Round**: 3,001 (15.0% Drop-off)
- **Offer Stage**: 1,691
- **Total Joined**: 1,470 (22.2% Overall Conversion Rate)


```python
import pandas as pd
import plotly.graph_objects as go

funnel_df = pd.DataFrame({
    'Stage': ['Applications', 'Screening', 'Interview', 'Technical Round', 'Offer Stage', 'Joined'],
    'Candidates': [6633, 5105, 4186, 3001, 1691, 1470],
    'Dropoff_Rate': [0.0, 23.0, 18.0, 28.9, 15.0, 13.0]
})

fig = go.Figure(go.Funnel(
    y=funnel_df['Stage'],
    x=funnel_df['Candidates'],
    textinfo="value+percent initial",
    marker=dict(color=["#38BDF8", "#60A5FA", "#EF4444", "#F59E0B", "#818CF8", "#22C55E"])
))
fig.update_layout(title='Recruitment Funnel Conversion Progression', template='plotly_dark')
fig.show()
```
